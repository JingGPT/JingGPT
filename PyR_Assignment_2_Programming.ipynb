# Assignment 2: Division... Without Dividing! 🍪

**The Challenge**

Imagine you have **11 cookies** and you want to pack them into boxes of **4**. How many full boxes do you get, and how many cookies are left over?

You could keep filling boxes (taking away 4 cookies each time) until you don't have enough cookies for another box. That's exactly what division is: **repeated subtraction**.

Your mission: write a program that divides **a** by **b** and finds the quotient (**q**) and remainder (**r**) <font color=red>**without using any division commands in Python** (no `/`, `//`, or `%`).</font>

> Subtract **b** from **a** again and again until you can't anymore. Count how many times you subtracted — that's your **quotient**. Whatever is left is your **remainder**.

**Assumptions (we're keeping it simple):**

*   a is always greater than b (a > b)
*   a and b are positive integers (a, b > 0)
*   a and b are not zero

**Example:**

> If a = 11 and b = 4:
>
> 11 − 4 = 7 ✅ (1 box packed)
>
> 7 − 4 = 3 ✅ (2 boxes packed)
>
> 3 − 4 = ? ❌ Can't do it — we stop!
>
> **Quotient = 2, Remainder = 3**

## Your Solution

Do **not** change the variable names `a`, `b`, `q`, or `r`, and do not change the print statements.

Complete the code so that when you change the values of `a` and `b`, your code correctly prints the **quotient** (q) and the **remainder** (r).

💡 **Before you type anything:** grab a pen and trace through a = 11, b = 4 by hand. If you can do it on paper, you can teach the computer to do it!

```python
a=11
b=4
# Write your code here:
a=int(input('enter first number'))
b=int(input('enter second number'))
q=0
r=a

while(r>=b):
  r=r-b
  q=q+1










#Do not change the print statements below
print("Quotient is",q)
print("Remainder is",r)
```
